State Description:

RobotAt(location_name)                          : The robot is at location location_name<br>
ItemAt(item_name,location_name)                 : The item item_name is at location location_name<br>

Connected(location_name_1,location_name_2)      : Location location_name_1 is connected to location location_name_2<br>

ContainerEmpty                                  : The robot has an empty container (no item inside)<br>
Carrying(item_name)                             : The robot is carrying the item item_name<br>

Delivered(item_name,location_name)              : The item item_name has been delivered to location location_name<br>

ElevatorAt(floor_no)                            : The elevator is at floor floor_no.<br>
InElevator                                      : The robot is inside the elevator<br>
ElevatorEmpty                                   : The elevator is empty (no robot inside)<br>

DoorClosed(location_name)                       : The door at guestroom location_name is closed<br>
Waiting(location_name)                          : The robot is waiting for hotel guest to open guest room door at location location_name<br>
DoorOpen(location_name)                         : The door at guestroom location_name is open<br>

Available item_name: Coffee, Water<br>

Connection: <br>

Elevator_F1 <-> Lobby <-> Hallway_F1 <-> Room101/Room102<br>
Elevator_F2 <-> Hallway_F2 <-> Room201/Room202<br>
